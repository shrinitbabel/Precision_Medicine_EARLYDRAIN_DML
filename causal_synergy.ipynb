```python
from modules.preprocess import load_and_clean_data
from modules.synergy import compute_synergy, plot_synergy, prepare_variables_exclude

df = load_and_clean_data("ed.csv")

treatment1 = "nimodipine"
treatment2 = "as_treated_flag" # LD or no LD
Outcome = "vs_clin"

# Drop nimodipine from X so it's not adjusted for
X, Y, T, feature_names = prepare_variables_exclude(df, outcome_col=Outcome, exclude_cols=[treatment1])

# Synergy between LD and nimodipine
synergy, summary = compute_synergy(df, X, Y, treatment2, treatment1, Outcome, feature_names)

print("🧪 Synergy Summary:")
for k, v in summary.items():
    print(f"{k}: {v:.4f}" if isinstance(v, float) else f"{k}: {v}")

plot_synergy(synergy, title="Synergy: LD + Nimodipine")
```
